```python
# mnist 
x,y = mnist["data"], mnist["target"]
# x.shape
# y.shape
```

```python
x_train, x_test, y_train, y_test = x.iloc[:60000], x.iloc[60000:], y.iloc[:60000], y.iloc[60000:]
```

```python
import numpy as np
shuffle_index = np.random.permutation(60000)
x_train, y_train = x_train.iloc[shuffle_index], y_train.iloc[shuffle_index]
```

## Creating a 2(Two) detector

```python
y_train = y_train.astype(np.uint8)
y_test = y_test.astype(np.uint8)
y_train_2 = (y_train == 2)
y_test_2 = (y_test == 2)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
clf = LogisticRegression()
clf.fit(x_train, y_train_2)
```

```python
clf.predict([some_digit])
```

```python
from sklearn.model_selection import cross_val_score
a = cross_val_score(clf, x_train, y_train_2, cv=3, scoring="accuracy")
```

```python
a.mean()
```

```python
from sklearn.model_selection import cross_val_predict


y_train_pred = cross_val_predict(clf, x_train, y_train_2, cv=3)
```

```python
y_train_pred
```

## Calculating confusion matrix

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_train_2, y_train_pred)

cm
```

```python
confusion_matrix(y_train_2, y_train_2)
# This is the confusion matrix for a perfect classifier. The diagonal elements are the number of correct predictions, while the off-diagonal elements are the number of incorrect predictions.
```

## Precision & Recall

```python
from sklearn.metrics import precision_score, recall_score, f1_score
```

```python
precision_score(y_train_2, y_train_pred)
# this is my precision score. It is the ratio of true positives to the sum of true positives and false positives. It is a measure of how many of the positive predictions were actually correct.
```

```python
recall_score(y_train_2, y_train_pred)
# this is my recall score. It is the ratio of true positives to the sum of true positives and false negatives. It is a measure of how many of the actual positive instances were correctly identified.
```

## F1 Score

```python
f1_score(y_train_2, y_train_pred)
# this is my F1 score. It is the harmonic mean of precision and recall. It provides a single metric that balances both concerns.
```

## Precision Recall Curve

```python
from sklearn.metrics import precision_recall_curve
```

```python
y_scores = cross_val_predict(clf, x_train, y_train_2, cv=3, method="decision_function")
```

```python
precisions, recalls, thresholds = precision_recall_curve(y_train_2, y_scores)
```

## Plotting the Precision Recall Curve

```python
plt.plot(thresholds, precisions[:-1], "b--", label="Precision")
plt.plot(thresholds, recalls[:-1], "g-", label="Recall")
plt.xlabel("Threshold")
plt.ylabel("Score")
plt.legend(loc="upper left")
plt.show()
```
